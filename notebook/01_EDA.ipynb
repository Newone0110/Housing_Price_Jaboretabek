```python
def bandingkan(i, df):
    r = df.loc[i]
    peers = df[(df["city"] == r["city"])
               & df["building_size_m2"].between(r["building_size_m2"]*0.8, r["building_size_m2"]*1.2)
               & df["land_size_m2"].between(r["land_size_m2"]*0.8, r["land_size_m2"]*1.2)
               & (df.index != i)]
    return len(peers), peers["price_in_rp"].median(), peers["price_in_rp"].min()

for i in df.nsmallest(10, "price_in_rp").index:
    print(i, df.loc[i, "price_in_rp"], bandingkan(i, df))
```

```python
df["hrg_m2_bgn"] = df["price_in_rp"] / df["building_size_m2"] / 1e6
med = df.groupby("district")["hrg_m2_bgn"].transform("median")
df["rasio"] = df["hrg_m2_bgn"] / med
df[df["rasio"] < 0.35][["title", "city", "district", "price_in_rp", "building_size_m2", "rasio"]]
```
